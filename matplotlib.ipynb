MATPLOTIB BASICS 

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 5 , 11) 
x
```

```python
y = x ** 2
y
```

```python
plt.plot(x,y)
plt.title("Square Numbers")
plt.xlabel(" X-Axis")
plt.ylabel("Y-Axis")
```

SUBPLOTS IN MATPLOTLIB  

```python
plt.subplot(2,2,1)
plt.plot(x**2,y**2)
plt.subplot(2,2,3)
plt.plot(x, y**3)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(y,x)
```

```python
# OBJECT ORIENTED WAY OF PLOTTING

fig = plt.figure(figsize=(8,4),dpi = 100)
axis1 = fig.add_axes([0.1,0.1,0.8,0.8]) # left, bottom, width, height
axis1.plot(x,y)
axis2 = fig.add_axes([0.1 , 0.1 , 0.8, 0.8])
axis2.plot(x , y)
```

```python
# TYPE OF PLOTS 

# 1. scatter plots 
plt.scatter(x,y)
```

```python
# 2. Histogram data 
from random import sample
data = sample(range(1,1000), 100)
plt.hist(data)
```

```python
# 3. box plot
plt.boxplot(x) 
```

```python
# SAVINGS PLOTS 

fig = plt.figure()
axis_1 = fig.add_axes([0.1,0.1,0.8,0.8])
axis_1.plot(x,y)
plt.savefig("plot.png")
```
